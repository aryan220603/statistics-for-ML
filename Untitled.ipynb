```python
## Z-score

import numpy as np

def detect_outliers(data):
    threshold = 2 # 3 standard deviations
    mean = np.mean(data)
    std = np.std(data)

    outliers = []

    if std == 0:
        return outliers

    for i in data:
        z_score = (i - mean) / std
        if abs(z_score) > threshold:
            outliers.append(i)

    return outliers
```

```python
data = [10, 12, 13, 15, 18, 20, 11100]

print(detect_outliers(data))
```
